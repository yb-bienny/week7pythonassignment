```python
# Import required libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Enable inline plotting if running in Jupyter Notebook
%matplotlib inline

# ----------------------------
# Task 1: Load and Explore the Dataset
# ----------------------------

try:
    # Load Iris dataset from sklearn and convert to DataFrame
    iris_raw = load_iris()
    iris = pd.DataFrame(iris_raw.data, columns=iris_raw.feature_names)
    iris['species'] = iris_raw.target
    iris['species'] = iris['species'].map(dict(enumerate(iris_raw.target_names)))

    # Display the first few rows
    print("First 5 rows of the dataset:")
    print(iris.head())

    # Check data types and missing values
    print("\nDataset Info:")
    print(iris.info())

    print("\nMissing values per column:")
    print(iris.isnull().sum())

    # Clean dataset (no missing values in Iris, but included for completeness)
    iris.dropna(inplace=True)

except FileNotFoundError:
    print("Error: The dataset file could not be found.")
except Exception as e:
    print(f"An error occurred: {e}")

# ----------------------------
# Task 2: Basic Data Analysis
# ----------------------------

# Basic statistics
print("\nBasic statistics of numerical columns:")
print(iris.describe())

# Group by species and compute mean
print("\nMean values grouped by species:")
grouped_means = iris.groupby('species').mean()
print(grouped_means)

# Pattern identification (example output)
print("\nInteresting Findings:")
print("• Setosa has shorter petal lengths on average.")
print("• Virginica has the highest sepal and petal dimensions.")

# ----------------------------
# Task 3: Data Visualization
# ----------------------------

# Set seaborn style
sns.set(style="whitegrid")

# 1. Line chart - Simulated time trend using index
plt.figure(figsize=(8, 5))
plt.plot(iris.index, iris['sepal length (cm)'], label='Sepal Length')
plt.plot(iris.index, iris['petal length (cm)'], label='Petal Length')
plt.title("Trend of Sepal and Petal Length Over Index")
plt.xlabel("Index")
plt.ylabel("Length (cm)")
plt.legend()
plt.show()

# 2. Bar chart - Average petal length per species
plt.figure(figsize=(6, 4))
sns.barplot(x='species', y='petal length (cm)', data=iris, palette='viridis')
plt.title("Average Petal Length per Species")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")
plt.show()

# 3. Histogram - Distribution of sepal width
plt.figure(figsize=(6, 4))
plt.hist(iris['sepal width (cm)'], bins=20, color='skyblue', edgecolor='black')
plt.title("Distribution of Sepal Width")
plt.xlabel("Sepal Width (cm)")
plt.ylabel("Frequency")
plt.show()

# 4. Scatter plot - Sepal vs. Petal Length
plt.figure(figsize=(6, 4))
sns.scatterplot(x='sepal length (cm)', y='petal length (cm)', hue='species', data=iris)
plt.title("Sepal Length vs. Petal Length by Species")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.legend(title="Species")
plt.show()
```

```python
Summary of Findings:
No missing values in the Iris dataset.

Setosa has distinctly lower petal length and width.

Virginica tends to have the largest measurements.

Scatter plots show clear clustering of species based on petal/sepal dimensions.
```
